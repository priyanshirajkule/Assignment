```python

import pandas as pd

# Create a sample dataset
data = {
    'Name': ['Aarav', 'Diya', 'Rohan', 'Meera', 'Diya'],
    'Marks': [85, 92, None, 76, 92],
    'City': ['Pune', 'Mumbai', 'Nashik', None, 'Mumbai'],
    'Remarks': ['Good', 'Excellent', 'Average', 'Good', 'Excellent']
}

df = pd.DataFrame(data)

# Display the original dataset
print("Original Dataset:")
print(df)

print("\nDataset Information:")
df.info()

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing marks with the mean
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())

# Fill missing cities with a placeholder
df['City'] = df['City'].fillna('Unknown')

print("\nDataset After Handling Missing Values:")
print(df)


# Remove duplicate rows
df = df.drop_duplicates()

print("\nDataset After Removing Duplicates:")
print(df)

# Remove an irrelevant column
df = df.drop(columns=['Remarks'])

print("\nDataset After Removing Irrelevant Column:")
print(df)

# Select specific columns using loc[]
print("\nSelected Columns:")
print(df.loc[:, ['Name', 'Marks']])

# Filter students scoring more than 80 marks
print("\nStudents Scoring More Than 80 Marks:")
print(df.loc[df['Marks'] > 80])

# Display the final cleaned dataset
print("\nFinal Cleaned Dataset:")
print(df)

```
