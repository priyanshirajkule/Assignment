```python
#list operations
l=[10,20,30,40,60,30]
print(l)
l.append(50)
print("after append:",l)
l.insert(2,45)
print("after insert:",l)
l.remove(45)
print("after remove:",l)
l.pop()
print("after pop:",l)
l.sort()
print("after sort",l)
```

```python
#tuple operations
t=(10,20,20,30,40,50)
print(t)
print("count of 20:", t.count(20))
print("index of 50:", t.index(50))
print("length of tuple:", len(t))
print("tuple slicing:",t[1:4])
```

```python
#dictionary operations
student={"student":"rahul","age":20,"course":"cse"}
print(student)
print("keys:", student.keys())
print("items:", student.items())
print("name:", student.get("name"))
student.update({"age":21})
print("updated dictionary:", student)
student.pop("course")
print("after pop", student)
```
